```python
import numpy as np
```

```python
# creating 2 matrices
# A m*n
# B n*p
A = np.array([[1, 2], [3, 4], [5, 6]])
B = np.array([[7, 8, 9], [10, 11, 12]])
```

```python
#A[row, column]
print(A[1,0])
print(A[0])
print(A[:,0]) # all rows, column 0
```

```python
#multiply A nd B
C = np.dot(A, B)
C
```

```python
A@B # same as np.dot(A, B)
```

```python
# Transpose of A
A.T
#or
np.transpose(A)
#or
A.transpose()
```

```python
# inverse can only be on a square matrix "np.linalg.inv(C)"
C = np.array([[1,2],[3,4]])
np.linalg.inv(C)

```

```python
# Determinant
np.linalg.det(C)
# if det =0 no inverse, if not equal to 0 invertible
```

```python
# rank
np.linalg.matrix_rank(A)
```

```python
# eigen values
np.linalg.eig(C)
```
